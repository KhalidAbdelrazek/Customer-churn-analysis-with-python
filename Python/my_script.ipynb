````python
# total charges outliers count
q1 = df['total_charges'].quantile(0.25)
q3 = df['total_charges'].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
outliers = df[(df['total_charges'] < lower_bound) | (df['total_charges'] > upper_bound)]
print(f"\nNumber of outliers in 'total_charges': {outliers.shape[0]}")
print(outliers[['customer_id', 'total_charges']])
print(f"q1: {q1}, q3: {q3}, iqr: {iqr}, lower_bound: {lower_bound}, upper_bound: {upper_bound}")
print(f"Min total_charges: {df['total_charges'].min()}, Max total_charges: {df['total_charges'].max()}")
````

````python
# percentile threshold
threshold = df['total_charges'].quantile(0.99)

# distribution of total_charges with threshold line
top_customers = df[df['total_charges'] >= threshold]


plt.figure(figsize=(6,4))
sns.histplot(df['total_charges'], kde=True, bins=50, color='blue', label='Total Charges')
plt.axvline(threshold, color='red', linestyle='--', label='99th Percentile Threshold')
plt.xlabel('Total Charges')
plt.title('Distribution of Total Charges with 99th Percentile Threshold')
plt.ylabel('Frequency')
plt.legend()
plt.savefig(os.path.join(figures_dir, "churn_by_total_charges.png"), dpi=300)
plt.show()

print(f"Threshold (99th percentile): {threshold}")
print(f"Number of customers in top 1%: {top_customers.shape[0]}")
print(top_customers[['customer_id', 'total_charges']].sort_values(by='total_charges', ascending=False).head(10))
````

````python
import plotly.express as px

# Create tenure categories
def categorize_tenure(months):
    if months <= 12:
        return "0-12 months"
    elif months <= 36:
        return "13-36 months"
    else:
        return "37+ months"

df["tenure_group"] = df["tenure"].apply(categorize_tenure)
df[["tenure", "tenure_group"]].head()
````

````python
# Step 2: Pie/Donut chart for customer distribution
tenure_counts = df["tenure_group"].value_counts()

fig, ax = plt.subplots(figsize=(6, 6))
ax.pie(
    tenure_counts,
    labels=tenure_counts.index,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops=dict(width=0.4)  # makes it a donut chart
)
ax.set_title("Customer Distribution by Tenure Group")
plt.savefig(os.path.join(figures_dir, "churn_by_tenure_group.png"), dpi=300)
plt.show()
````

````python
# Clustered bar chart for average monthly charges by tenure group
avg_charges = df.groupby("tenure_group")["monthly_charges"].mean().reset_index()

fig = px.bar(
    avg_charges,
    x="tenure_group",
    y="monthly_charges",
    text="monthly_charges",
    color="tenure_group",
    title="Average Monthly Charges by Tenure Group"
)
fig.update_traces(texttemplate='%{text:.2f}', textposition="outside")
plt.savefig(os.path.join(figures_dir, "churn_by_tenure_group.png"), dpi=300)
fig.show()
````

# Task 5: Advanced Analysis

## 📌 Description
This task focuses on performing deeper customer analytics by exploring relationships between **tenure, charges, churn, and demographics**. The goal is to uncover hidden patterns that explain churn behavior and customer lifecycle trends.

## 🎯 Objectives

## 📊 Tools
- **Pandas** – for grouping, aggregating, and calculating statistics.
- **Matplotlib / Seaborn** – for static visualizations.
- **Plotly** – for interactive exploration of churn trends.


````python
# ---- Grouping by tenure to compute statistics ----
tenure_stats = df.groupby('tenure').agg(
    avg_monthly_charges=('monthly_charges', 'mean'),
    avg_total_charges=('total_charges', 'mean'),
    churn_rate=('churn', lambda x: (x == 'Yes').mean())
).reset_index()

print("Tenure Statistics:\n", tenure_stats.head())

plt.figure(figsize=(8, 4))
sns.lineplot(data=tenure_stats, x='tenure', y='churn_rate', marker="o")
plt.title("Churn Rate by Tenure")
plt.xlabel("Tenure (months)")
plt.ylabel("Churn Rate")
plt.savefig(os.path.join(figures_dir, "churn_by_tenure.png"), dpi=300)
plt.show()
````

````python
demographic_cols = ['gender', 'senior_citizen', 'payment_method', 'contract']

for col in demographic_cols:
    churn_rates = (
        df.groupby(col)['churn']
        .value_counts(normalize=True)
        .unstack()
        .fillna(0)[['No', 'Yes']]  # force order
    )
    
    ax = churn_rates.plot(
        kind='bar', 
        stacked=True, 
        figsize=(6,4), 
        color=['#66c2a5', '#fc8d62']
    )
    
    plt.title(f"Churn Rate by {col.capitalize().replace('_',' ')}")
    plt.ylabel("Proportion")
    plt.xlabel(col.capitalize().replace('_',' '))
    plt.legend(title="Churn", loc="upper right")
    plt.savefig(os.path.join(figures_dir, f"churn_by_{col}.png"), dpi=300)
    plt.show()
````

````python
# ---- Lifecycle / contract type analysis ----
plt.figure(figsize=(6,4))
sns.boxplot(data=df, x='contract', y='monthly_charges', hue='churn', palette="Set2")
plt.title("Monthly Charges by Contract Type and Churn")
plt.xlabel("Contract Type")
plt.ylabel("Monthly Charges")
plt.legend(title="Churn")
plt.savefig(os.path.join(figures_dir, "churn_by_contract.png"), dpi=300)
plt.show()
````

````python
# relations between churn and {phone_service, multiple_lines, internet_service, online_security, online_backup, device_protection, tech_support, streaming_tv, streaming_movies}
for col in ['phone_service', 'multiple_lines', 'internet_service', 'online_security', 'online_backup', 'device_protection', 'tech_support', 'streaming_tv', 'streaming_movies']:
    plt.figure(figsize=(6,4))
    sns.countplot(data=df, x=col, hue='churn', palette="Set2")
    plt.title(f"{col} vs Churn")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.legend(title="Churn", loc="upper right")
    plt.savefig(os.path.join(figures_dir, f"churn_by_{col}.png"), dpi=300)
    plt.show()
````

````python
# Features to plot
features = [
    'phone_service', 'multiple_lines', 'internet_service',
    'online_security', 'online_backup', 'device_protection',
    'tech_support', 'streaming_tv', 'streaming_movies'
]

# Set up grid (3 rows x 3 cols for 9 features)
fig, axes = plt.subplots(3, 3, figsize=(18, 12))
axes = axes.flatten()

# Loop over features & plot on subplots
for i, col in enumerate(features):
    sns.countplot(data=df, x=col, hue='churn', palette="Set2", ax=axes[i])
    axes[i].set_title(f"{col} vs Churn")
    axes[i].set_xlabel(col)
    axes[i].set_ylabel("Count")
    axes[i].legend(title="Churn", loc="upper right")

# Adjust layout
plt.tight_layout()

# Save as one figure
plt.savefig(os.path.join(figures_dir, "churn_by_services.png"), dpi=300)
plt.show()
````

````python
import matplotlib.pyplot as plt

# Churn counts
churn_counts = df['churn'].value_counts()
labels = churn_counts.index
sizes = churn_counts.values

# Custom format: percent + count
def autopct_format(pct, all_vals):
    absolute = int(round(pct/100.*sum(all_vals)))
    return f"{pct:.1f}% ({absolute})"

# Plot
plt.figure(figsize=(8, 8))
wedges, texts, autotexts = plt.pie(
    sizes,
    labels=None,  # hide labels on slices
    autopct=lambda pct: autopct_format(pct, sizes),
    startangle=90,
    colors=['#66c2a5', '#fc8d62'],
    wedgeprops=dict(width=0.4),
    pctdistance=1.2,   # place values outside
)

# Style the percentage+count labels
for autotext in autotexts:
    autotext.set_fontsize(11)
    autotext.set_color("black")
    autotext.set_weight("normal")

# Add legend
plt.legend(
    wedges, labels,
    title="Churn Status",
    loc="center left",
    bbox_to_anchor=(1, 0, 0.5, 1),
    fontsize=12
)

# Title
plt.title("Customer Churn Distribution", fontsize=16, pad=20)
# save figure 
plt.savefig(
    os.path.join(figures_dir, "churn_distribution_pie_chart.png"),
    dpi=500,
    bbox_inches="tight"   # ensures legend & title are included
)
plt.show()
````
